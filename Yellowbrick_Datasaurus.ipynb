Back in 2018, Larry Gray, PhD infused the [Yellowbrick](https://www.scikit-yb.org/) machine learning diagnostic visualization library with the importance of visualization when he [added core Datasaurus datasets](https://github.com/DistrictDataLabs/yellowbrick/issues/257) to it, as part of his many contributions to the library as one of its primary project maintainers. Yellowbrick's maintainers believe that visual diagnostics are vital to machine learning.

Yellowbrick, which is built on Matplotlib, has extended the Scikit-Learn API to, "to make model selection and hyperparameter tuning easier," [YB](https://www.scikit-yb.org/en/latest/about.html) since 2016.

But Datasaurus illustrates a phenomena of visual data science that traces its roots much earlier to 1973.

Read more about the background of this phenomenum of visual data science from [Francis John Anscombe's quartet](https://en.wikipedia.org/wiki/Anscombe%27s_quartet) published in ["Graphs in Statistical Analysis"](https://www.jstor.org/stable/2682899) in _The American Statistician_ in 1973, to Alberto Cairo's [popular tweet circa 2016](https://twitter.com/albertocairo/status/765167969139765250) that introduced his Datasaurus dataset interpretation detailed in a [blog post on _thefunctionalart.com_](http://www.thefunctionalart.com/2016/08/download-datasaurus-never-trust-summary.html) that same year, and finally the modernization [The Datasaurus Dozen](https://www.autodesk.com/research/publications/same-stats-different-graphs), actually 13 datasets, introduced by Justin Matejka and George Fitzmaurice through the ACM SIGCHI Conference on Human Factors in Computing Systems in 2017.

John W Tukey, another contemporary statistician of Anscome's, who was also his brother-in-law, summed up this lesson of the importance of not just summarizing but also graphing your data best when he said in his book ["Exploratory Data Analysis"](https://www.worldcat.org/title/exploratory-data-analysis/oclc/3058187).

> The greatest value of a picture is when it forces us to notice what we never expected to see.

Follow along to [try Anscombe's Quartet for yourself](https://www.scikit-yb.org/en/latest/api/anscombe.html) with Yellowbrick!

```python
# optionally list all installed python modules in environment to verify what will be recognized for imports
# help('modules')
```

```python
# Imports

import yellowbrick as yb
# Pyplot sub-module of Matplotlib that allows you to make simple 2D plots
import matplotlib.pyplot as plt
# Python Imaging Library (PIL) Image module allows you to convert a PNG file to JPEG 
from PIL import Image
```

```python
# make a built-in grid plot of Anscombe’s Quartet[1] as an illustration of the importance of visualization
g = yb.anscombe()
plt.show()
```

```python
# make a 2x2 grid plot using Matejka and Fitzmaurice's Datasaurus Dozen dataset which is built into Yellowbrick
g = yb.datasaurus()
plt.show()
```

# Export the Datasaurus image for a blog post

First you will export the plot to an image file and then optimize it for web viewing to be published in a resolution quality suitable for fast loading on the web at 72 DPI (dots per inch).

JPEG (Joint Photographic Experts Group) images are suitable for the web, but are not a supported file format here.[3]. You will need to convert the image once it is saved to a PNG (Portable Network Graphic) file to JPEG. Due to the differences in these image formats–PNG which uses RGBA meaning Red, Green, Blue, plus Alpha for transparency and JPEG which uses just RGB—you will also need to remove the transparency channel before converting to JPEG which does not support transparency.

```python
# make a folder in your project directory with the common nomenclature 'img' to save static images in
# using a bash command for make directory 'mkdir' 
# preceded by the prefix '!' to run it with iPython right here in a Jupyter Notebook
!mkdir img
```

```python
# call the datasauras function again, but this time you won't bother showing it interactively in the notebook output
g = yb.datasaurus()

# export the figure containing the plot you made with the datasaurus data as PNG in the img folder you just made
# set dpi, and set the bounding box for the portion of the figure to be saved
plt.savefig('img/Datasaurus_from_Yellowbrick.png', dpi=72, bbox_inches='tight')

# open the PNG image file and save it as a variable object named 'im'
im = Image.open('img/Datasaurus_from_Yellowbrick.png')
# convert RGBA color to RGB to remove alpha transparency channel
rgb_im = im.convert('RGB')

# save a converted image file with a specified JPEG format and extension in the same location.
rgb_im.save('img/Datasaurus_from_Yellowbrick.jpg','JPEG')
```
